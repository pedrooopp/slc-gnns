```python
import matplotlib.pyplot as plt
import re
import numpy as np

# Align both DataFrames (important!)
df1_aligned, df2_aligned = df_new.align(df_old, join="inner", axis=0)  # align index
df1_aligned, df2_aligned = df1_aligned.align(df2_aligned, join="inner", axis=1)  # align columns

# Loop through columns and plot
for col in df1_aligned.columns:

    var_name = re.sub(r"\([^)]*\)|\[[^\]]*\]|\{[^}]*\}", "", col[0])
    var_name = re.sub(r"\s+", " ", var_name).strip()

    unit = re.search(r"\[([^\]]*)\]", col[0])

    if unit:
        unit = unit.group(1)
    else:
        raise ValueError(f"No se encontro unidad para generar el plot ({col[0]})!")

    plt.figure(figsize=(10, 6))
    
    plt.xlabel(f"Log {var_name} nuevo [{unit}]")
    plt.ylabel(f"Log {var_name} viejo [{unit}]")
    plt.title(f"Diagrama de disperisón de '{var_name}' - Estación '{col[1]}'")
    
    # Optional: 1:1 reference line
    min_val = min(df1_aligned[col].min(), df2_aligned[col].min())
    max_val = max(df1_aligned[col].max(), df2_aligned[col].max())
    plt.plot([min_val, max_val], [min_val, max_val], linestyle='--', color="black", zorder=0)

    plt.scatter(df1_aligned[col], df2_aligned[col], s=10)

    plt.xscale("log")
    plt.yscale("log")


    # x = df1_aligned[col].dropna()
    # y = df2_aligned[col].dropna()

    # # Compute limits ignoring extreme values
    # low = min(np.percentile(x, 5), np.percentile(y, 5))
    # high = max(np.percentile(x, 95), np.percentile(y, 95))

    # plt.xlim(low, high)
    # plt.ylim(low, high)
    
    plt.savefig(f'scatter_plots/logs_plot_{var_name}_{col[1].replace(" ","_")}.png')
    plt.close()
```
