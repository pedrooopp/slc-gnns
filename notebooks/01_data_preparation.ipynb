```python
import re

def extract_variable(col):
    """
    Extract variable name from:
    [Station] Variable
    """

    match = re.search(
        r"\]\s*(.*)",
        col
    )

    if match:
        return match.group(1).strip()

    return col

import pandas as pd

def missing_by_variable(df):

    # Missing % per column
    missing_col = (
        df.isna().mean() * 100
    )

    # Extract variable names
    variable_names = (
        missing_col.index
        .map(extract_variable)
    )

    # Group by variable
    missing_var = (
        missing_col
        .groupby(variable_names)
        .mean()
        .sort_values()
    )

    return missing_var

import matplotlib.pyplot as plt

def plot_missing_variable(df, title, path):

    sns.set_theme(style="whitegrid")

    missing_var = missing_by_variable(df)

    # Optional: sort values (recommended for horizontal bars)
    missing_var = missing_var.sort_values(ascending=False)

    plt.figure(figsize=(16, 10))

    ax = sns.barplot(
        x=missing_var.values,
        y=missing_var.index,
        hue=missing_var.index,   # Fix for palette warning
        orient="h",
        palette="viridis"
    )
    ax.tick_params(axis="y", labelsize=14)
    ax.tick_params(axis="x", labelsize=14)
    ax.set_xlim(0, 100)

    for i, v in enumerate(missing_var.values):
        ax.text(v + 0.5, i, f"{v:.0f}%", va="center", fontsize=13)

    plt.xlabel("Missing Data (%)", fontsize=14)
    plt.ylabel("")

    plt.title(
        title, fontsize=16
    )

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300
    )

    plt.close()

#plot_missing_variable(df_dinacea_bimcustom)
```

```python
import re

def extract_variable(col):
    """
    Extract variable name from:
    [Station] Variable
    """

    match = re.search(
        r"\]\s*(.*)",
        col
    )

    if match:
        return match.group(1).strip()

    return col


def create_variable_groups(df):

    groups = {}

    for col in df.columns:

        var = extract_variable(col)

        if var not in groups:
            groups[var] = []

        groups[var].append(col)

    return groups

groups = create_variable_groups(df_inumet)
```

```python
import os
import re
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def extract_var_and_unit(var):

    # Remove text inside ()
    var_name = re.sub(r"\([^)]*\)", "", var)

    # Normalize spaces
    var_name = " ".join(var_name.split())

    # Extract unit from last []
    unit_match = re.findall(r"\[([^\]]+)\]", var)

    unit = unit_match[-1] if unit_match else ""

    return var_name, unit

def plot_variable_boxplots(df, groups, path):

    os.makedirs("boxplots", exist_ok=True)

    # Nice default seaborn style
    sns.set_theme(
        style="whitegrid",
        context="talk"   # bigger text
    )

    for var, cols in groups.items():

        var_name, unit = extract_var_and_unit(var)

        ylabel = f"{var_name} [{unit}]"


        # --- Reshape to long format ---
        df_long = (
            df[cols]
            .melt(
                var_name="Station",
                value_name="Value"
            )
            .dropna()
        )

        df_long = df_long[df_long["Value"] > 0]

        # Clean station names
        df_long["Station"] = (
            df_long["Station"]
            .str.extract(r"\[([^\]]+)\]")
            .iloc[:, 0]
        )

        # --- Create figure ---
        fig, ax = plt.subplots(
            figsize=(max(12, len(cols) * 0.6), 8)
        )

        # --- Boxplot ---
        sns.boxplot(
            data=df_long,
            x="Station",
            y="Value",
            ax=ax,
            width=0.6,
            fliersize=2,
            linewidth=1,
            palette="viridis"
        )
        ax.set_yscale("log")


        # --- Formatting ---
        ax.set_title(
            f'Boxplot for variable "{var}" (log without zeros)',
            fontsize=15,
            pad=15
        )

        ax.set_xlabel(
            "Station",
            fontsize=13
        )

        ax.set_ylabel(
            ylabel,
            fontsize=13
        )

        ax.tick_params(
            axis="x",
            rotation=90,
            labelsize=10
        )

        ax.tick_params(
            axis="y",
            labelsize=11
        )

        sns.despine()

        plt.tight_layout()

        # --- Safe filename ---
        safe_var = re.split(r"[\(\[]", var)[0].strip()

        filename = (
            path
        ).replace(" ", "_")

        plt.savefig(
            f"{filename}_{safe_var}.png",
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

plot_variable_boxplots(df_inumet, groups, path=f"boxplots/03_box_inumet_log")
```

```python
import os
import re
import calendar
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


def plot_seasonal_by_variable(df, groups, path):

    os.makedirs("seasonal", exist_ok=True)

    # Global style
    sns.set_theme(
        style="whitegrid",
        context="talk"   # larger fonts
    )

    for var, cols in groups.items():

        df_var = df[cols]

        # --- Monthly aggregation ---
        monthly = df_var.groupby(
            df_var.index.month
        ).median()

        # Median across stations
        monthly_median = monthly.median(axis=1)

        # IQR for shading
        q25 = monthly.quantile(0.25, axis=1)
        q75 = monthly.quantile(0.75, axis=1)

        # Month labels
        month_labels = [
            calendar.month_abbr[m]
            for m in monthly.index
        ]

        # --- Plot ---
        fig, ax = plt.subplots(figsize=(10, 6))

        sns.lineplot(
            x=monthly.index,
            y=monthly_median,
            marker="o",
            linewidth=2.5,
            ax=ax
        )

        # Shaded IQR band
        ax.fill_between(
            monthly.index,
            q25,
            q75,
            alpha=0.2
        )

        # Labels
        ax.set_title(
            f"Seasonal Pattern — {var}",
            fontsize=16,
            pad=12
        )

        ax.set_xlabel(
            "Month",
            fontsize=13
        )

        ax.set_ylabel(
            "Median",
            fontsize=13
        )

        # Month names instead of numbers
        ax.set_xticks(monthly.index)
        ax.set_xticklabels(month_labels)

        # Slight margins
        ax.margins(x=0.02)

        sns.despine()

        plt.tight_layout()
        print(var)

        # --- Safe filename ---
        safe_var = re.split(r"[\(\[]", var)[0].strip()

        filename = (
            f"{path}_{safe_var}.png"
        ).replace(" ", "_")

        plt.savefig(
            filename,
            dpi=300
        )

        plt.close()


plot_seasonal_by_variable(
    df_inumet,
    groups, path=f"seasonal/04_seasonal_dinagua"
)
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

def create_missing_data_plot(data, save_path, station, start_date, end_date):

    present_color = "#2CA02C"  # Dark Green
    missing_color = "#E74C3C"  # Orange-Red

    # Create figure
    fig = go.Figure()

    variables = data.columns

    data = data.reset_index(names="time")

    if data[data["time"]<start_date].empty: # Is there an entry older than start_date?
        # Add a new empty row --> Indicates empty data on the plot.
        start_date_row = pd.DataFrame({'time': [pd.Timestamp(start_date)]}) # Format: 2024-01-01
        data = pd.concat([start_date_row, data], ignore_index=True)

        
    if data[data["time"]>end_date].empty: # Is there an entry newer than end_date?
        # Por como se traza la linea roja, necesito agregar un elemento vacio al dia siguiente
        end_date_row_delta = pd.DataFrame({'time': [pd.Timestamp(data["time"].max()+pd.Timedelta(days=1))]}) 
        end_date_row = pd.DataFrame({'time': [pd.Timestamp(end_date)]}) 
        data = pd.concat([data, end_date_row_delta, end_date_row], ignore_index=True)

    # Process each variable separately
    for var in variables:
        
        df_var = data.loc[:,["time", var]].copy()
        df_var["is_missing"] = df_var[var].isna()
        df_var["group"] = (df_var["is_missing"] != df_var["is_missing"].shift()).cumsum()
        df_var["variable"] = var

        for _, group in df_var.groupby("group"):

            # El grupo contiene el ultimo indice? --> Si no, puedo "agarrar" la fecha del siguiente cambio (dato faltante/no faltante)
            if group.index[-1] != df_var.index[-1]: 


                new_row = df_var.loc[group.index[-1] + 1].to_frame().T
                new_row = new_row.astype(group.dtypes)
                group = pd.concat([group, new_row], ignore_index=True)
                group.loc[len(group) - 1, "is_missing"] = not group.loc[len(group) - 1, "is_missing"]

                color = missing_color if group["is_missing"].iloc[0] else present_color
                fig.add_trace(go.Scatter(
                    x=group["time"], y=group["variable"],
                    mode="lines",
                    line=dict(color=color, width=7),
                    name=f"{var} ({'Missing' if color == missing_color else 'Present'})",
                    showlegend=False
                ))
            # El grupo contiene el ultimo indice y tiene una sola fila --> Hay un cambio (dato faltante/no faltante). Agrego una punta de flecha al ultimo día
            elif group.shape[0] == 1:
                color = missing_color if group["is_missing"].iloc[0] else present_color
                fig.add_trace(go.Scatter(
                    x=[group["time"].iloc[-1]], y=[group["variable"].iloc[-1]],
                    mode="markers",
                    marker=dict(symbol="diamond", color=color, size=16),
                    name=f"{var} ({'Missing' if color == 'red' else 'Present'})",
                    showlegend=False
                ))
            # El grupo contiene el ultimo indice y no tiene una sola fila --> NO hay un cambio (dato faltante/no faltante). Continuo la recta del mismo color (roja/verde)   
            else:
                # Ajusto la fecha del comienzo considerando "time_tuning_for_arrow"
                # Esto resulta en solamente tener un cambio de "time_tuning_for_arrow" cuando hay un cambio entre dato faltante/no faltante en el ultimo dato
                # Se hace esto para poder utilizar una flecha sin que se superponga una linea del color opuesto.
                color = missing_color if group["is_missing"].iloc[0] else present_color
                fig.add_trace(go.Scatter(
                    x=group["time"], y=group["variable"],
                    mode="lines",
                    line=dict(color=color, width=7),
                    name=f"{var} ({'Missing' if color == missing_color else 'Present'})",
                    showlegend=False
                ))

    # Layout
    fig.update_layout(
        xaxis_title="Tiempo",
        yaxis_title="Variable",
        yaxis=dict(
            categoryorder="category ascending", 
            autorange=True, 
            fixedrange=False, 
            tickfont=dict(size=20),
            title_font=dict(size=24),
        ),
        title=dict(
            text=f"Evolución temporal de la disponibilidad de datos por variable (estación {station})",
            font=dict(size=29),
            x=0.5  # Center the title
        ),
        xaxis=dict(
            title_font=dict(size=24),
            tickfont=dict(size=20),
            tickformat="%Y"  # Improved timestamp formatting
        ),
        width=2000,
        height=600,
        )

    fig.update_xaxes(range=[start_date, end_date])

    fig.write_image(f"{save_path}_missing_data.png")     
```
